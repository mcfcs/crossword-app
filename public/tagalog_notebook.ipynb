```python
import re
import pandas as pd

def remove_exact_word_from_clue(row):
    w = str(row["word"]).strip()
    c = str(row["definition"])

    if not w or pd.isna(row["definition"]):
        return row["definition"]

    # match Word as a whole token (not inside other words), case-insensitive
    pat = re.compile(rf"(?i)(?<![A-Za-z0-9]){re.escape(w)}(?![A-Za-z0-9])")

    # remove (optionally remove only the first occurrence: count=1)
    c = pat.sub("", c, count=1)

    # cleanup: extra spaces + leading punctuation left behind
    c = re.sub(r"\s{2,}", " ", c).strip()
    c = re.sub(r"^[\s,;:\-–—]+", "", c).strip()

    return c

df["definition"] = df.apply(remove_exact_word_from_clue, axis=1)
```

```python
import pandas as pd
import re
from difflib import SequenceMatcher

df0 = df.copy()  # keep original safe

# -----------------------
# Similarity filter
# -----------------------
def _norm(s: str) -> str:
    # normalize for comparisons (remove punctuation/hyphens/spaces, lowercase)
    return re.sub(r"[^A-Za-z0-9]+", "", str(s)).casefold()

def too_similar(word: str, clue_word: str, min_root=4, ratio_thresh=0.85) -> bool:
    w, c = _norm(word), _norm(clue_word)
    if not w or not c:
        return False
    if w == c:
        return True

    # Reject prefix/containment like TALA vs TALAAN
    short = min(len(w), len(c))
    if short >= min_root and (w.startswith(c) or c.startswith(w) or w in c or c in w):
        return True

    # Reject high similarity like ITAHI vs IPAGTAHI, SOOLOHIA vs SOOLOHIYA
    if SequenceMatcher(None, w, c).ratio() >= ratio_thresh:
        return True

    return False

# -----------------------
# Base cleaned view (no deletion from df0)
# -----------------------
base = (
    df0[["Date", "Word", "Clue"]]
      .dropna(subset=["Word", "Clue"])
      .assign(
          Word=lambda x: x["Word"].astype(str).str.strip(),
          Clue=lambda x: x["Clue"].astype(str).str.strip(),
      )
)

# Find which clues are unique vs shared
clue_counts = base.groupby("Clue")["Word"].nunique()
unique_clue_values = clue_counts[clue_counts == 1].index
shared_clue_values = clue_counts[clue_counts > 1].index

# -----------------------
# uniqueclues: keep original rows with unique clue
# -----------------------
uniqueclues = base[base["Clue"].isin(unique_clue_values)].copy()

# -----------------------
# synonymclues: for shared clues, generate Word->Word clue rows
# -----------------------
shared = (
    base[base["Clue"].isin(shared_clue_values)]
      .drop_duplicates(subset=["Clue", "Word"])  # avoid duplicates inside a clue-group
)

def make_synonym_rows(g: pd.DataFrame) -> pd.DataFrame:
    words = g["Word"].unique().tolist()
    date_by_word = g.drop_duplicates("Word").set_index("Word")["Date"]

    rows = []
    for w1 in words:
        for w2 in words:
            if w1 == w2:
                continue
            if too_similar(w1, w2):
                continue
            rows.append({"Date": date_by_word[w1], "Word": w1, "Clue": w2})
    return pd.DataFrame(rows)

synonymclues = (
    shared.groupby("Clue", group_keys=False)
          .apply(make_synonym_rows)
          .drop_duplicates(subset=["Word", "Clue"])
          .reset_index(drop=True)
)

# -----------------------
# uniqueclues: UNIQUE WORDS (no Tagalog synonym rows created)
# -----------------------
words_with_synonyms = set(synonymclues["Word"])

uniqueclues = base[~base["Word"].isin(words_with_synonyms)].copy()
# (optional) remove duplicate original rows
uniqueclues = uniqueclues.drop_duplicates(subset=["Word", "Clue"]).reset_index(drop=True)

# -----------------------
# allclues = synonymclues + uniqueclues
# -----------------------
allclues = (
    pd.concat([synonymclues, uniqueclues], ignore_index=True)
      .drop_duplicates(subset=["Word", "Clue"])
      .reset_index(drop=True)
)

```

```python
import pandas as pd

# Base original rows (word + original clue)
base = (
    df[["Date", "Word", "Clue"]]
      .dropna(subset=["Word", "Clue"])
      .assign(
          Word=lambda x: x["Word"].astype(str).str.strip(),
          Clue=lambda x: x["Clue"].astype(str).str.strip(),
      )
      .drop_duplicates(subset=["Word", "Clue"])
)

# Rows whose clue appears for only ONE word (no synonyms possible)
clue_counts = base.groupby("Clue")["Word"].nunique()
unique_clues = clue_counts[clue_counts == 1].index
unique_rows = base[base["Clue"].isin(unique_clues)]

# Only keep those words if they are NOT already present in synonym_clues
unique_rows = unique_rows[~unique_rows["Word"].isin(allclues["Word"])]

# Append + de-dupe
df_clues = (
    pd.concat([allclues, unique_rows], ignore_index=True)
      .drop_duplicates(subset=["Word", "Clue"])
      .reset_index(drop=True)
)

# Clue: capitalize first letter only
df_clues["Clue"] = (
    df_clues["Clue"].astype(str).str.strip().str.lower().str.capitalize()
)
df_clues
```

```python
import re

def _alnum(s: str) -> str:
    return re.sub(r"[^A-Za-z0-9]+", "", str(s)).casefold()

def word_is_in_clue(row, min_len=3):
    w = _alnum(row["Word"])
    c = _alnum(row["Clue"])
    if len(w) < min_len:
        return False
    return w in c

df_clues = df_clues[~df_clues.apply(word_is_in_clue, axis=1)].reset_index(drop=True)
```

```python
import re
from difflib import SequenceMatcher

def _norm(s: str) -> str:
    # normalize: keep letters/numbers only, lowercase
    return re.sub(r"[^A-Za-z0-9]+", "", str(s)).casefold()

def similarity(a: str, b: str) -> float:
    a, b = _norm(a), _norm(b)
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, a, b).ratio()

# tune this threshold
THRESH = 0.60  # try 0.80–0.90 depending on how aggressive you want

mask_too_similar = df_clues.apply(
    lambda r: similarity(r["Word"], r["Clue"]) >= THRESH,
    axis=1
)

df_clues = df_clues[~mask_too_similar].reset_index(drop=True)
df_clues
```
