```python
import pandas as pd

url = "https://tl.wikipedia.org/wiki/Talaan_ng_mga_lungsod_at_bayan_sa_Pilipinas"

# This returns a list of DataFrames for ALL tables on the page
tables = pd.read_html(url)

# On this page, the big “Talaan” table is typically the largest one
df = max(tables, key=lambda t: t.shape[0])

print(df.shape)
print(df.head())
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

url = "https://tl.wikipedia.org/wiki/Talaan_ng_mga_lungsod_at_bayan_sa_Pilipinas"
html = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}).text
soup = BeautifulSoup(html, "html.parser")

table = soup.select_one("table.wikitable.sortable.plainrowheaders.mw-datatable")
df = pd.read_html(str(table))[0]

print(df.shape)
print(df.head())
```

```python
df_cleaned = df.drop(columns=["Populasyon (2015)[1]", "Lawak/sukat (km2)[i][2][3]", "Barangay", "Klase", "Unnamed: 7", "Unnamed: 8", "Unnamed: 9"] )
df_cleaned
```
